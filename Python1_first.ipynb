### Functions

```python
# we declare reuseable code in a function
def showDetails(n):
    '''here we iterate over every member of collection n
    We can have several lines in a docstring (triple quotes)
    if the collection is ordinal, just iterate
    If the collection is a dict, show k:v pairs'''
    # check if dict
    if type(n) is dict:
        for (k,v) in n.items():
            print(k, v)
    else:
        for i in n:
            print(i)

# we then invoke the function whenever we need it
showDetails(me)
showDetails(tickers)
showDetails(hello)
showDetails(details)
```
