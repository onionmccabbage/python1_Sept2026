## Dictionaries & Tuples

#### Dictionaries
Slide 33

```python
# Storing information in key:value pairs
stocks = {'AAPL':'Apple Inc.', 
		  'CAT':'Caterpillar', 
		  'MSFT':'Microsoft',
          'NFLX':'Netflix'}

#To extract
stocks['CAT']

#To change values
stocks['AAPL'] = "Apple"

#To create a new entry
stocks['FB'] = 'Facebook'

stocks
```

```python
#list of all row names
stocks.keys()
#list of all the values
stocks.values()
```

```python
# iterate over members of a dict
for (key, value) in stocks.items():
    print(f'{key}: {value}')
    # order doesn't matter
    # print(f'{value}: {key}')
```

#### If Statements - Boolean Logic
Slides 36 - 40

```python
# or |
# and &
#not equal to: !=
#to compare if two things are equal use == rather than just =
#one "=" is used to change the value of a variable

x = 5
y = 10

# y = x #this changes b to be equal to x
# x > 0 & y <100
# x > 0 and y <100
y == x #this checks if y and x are the same
```

```python
#If Example
if x > y:
    print("x is greater than y")
    print("x is ", x)
    if x > 100:
        print("x is greater than 100")
elif y > x:
    print("y is greater than x")
else:
    print("x is equal to y")

print("this is outside the if statement")
```

```python
#Be careful with "Truthy" statements, anything that is not 0 is True
if 10 - 2:
    print("This IF is True")
```

#### For Loop
Slide 41    

__Looping__

* Two main types of loops: For loop, While loop
* For Loop - you know how many times to loop
* While Loop - you are checking a condition
* A while loop will run as long as condition is True

```python
#Looping through a list
numbers = [2, 5, 7, 9, 13, 25]
#Print the squares and cubes of these numbers really quickly
for num in numbers:
    # print("-"*10)
    print(f"{num} square: {num ** 2} Cube: {num ** 3}") 
```

```python
#Looping "X" number of times
for i in range(5): #prints numbers 0, 1, 2, ... 4
    print(i)

for x in range(2,26,2): #range(start, end, skip)
    print(x, end=', ')
print() #prints 2, 4, 6, etc. stops at 24
```

#### List Comprehension

```python
nums = [5, 7, 9, 12]
numsCube = [x**3 for x in nums]
numsCube
```

```python
#Faster than doing:
cubeList = []
for x in nums:
    cubeList.append(x**3)
```

```python
#Can combine with IF Statements
nums = [3, 50, 29, 12, 100, 62]
squareList = [x**2 for x in nums if x > 50]
squareList
```

#### While Loops
Slide 43

```python
x = 5
#this is an infinite loop, it never stops
# you can interrupt by hitting CTRL + C
#while x > 2:
#    print(x)
        
while x < 10: # originally 1000
    print(x)        
    x += 1  #x = x + 1
```

```python
#Careful with using floats in If statements or While loops (or any logic)
print(1.1 + 2.2 == 3.3) #Gives False
        #actually doing 3.30000001 or 3.29999999998...
print("{:.20f}".format(3.3)) # to 20 dp

round(1.1 + 2.2)  == round(3.3) #Gives True
```
