#### Strings and String Methods

```python
y = '4' # this is a string of characters
print(y * 100) # will print '4' repeated 100 times
print("-" * 50) #repeats the dash 50 times
```

```python
# Cleaning up numbers
myNumber = "$123,456,789"
myNumber * 2 # not what we wanted
```

```python
myNumber.strip('$') # did not change original
myNumber = myNumber.strip('$')
myNumber = myNumber.replace(',', '')
myNumber = int(myNumber)
myNumber # careful not to run this cell twice...
```

```python
#All in one step
myNumber = "$123,456,789"
myNumber = int(myNumber.strip('$').replace(',', ''))
myNumber
#variableName --> camel case
#variable_name --> snake case
```

#### String Format Method
Slide 23

```python
output = "Sales: {}, Gross Margin {:.2%}"
print(output.format(2000,0.3))

output2 = "Market Cap: {:>9,}"
mkCap1 = 1234
mkCap2 = 3456700
mkCap3 = 100234
print(output2.format(mkCap1))
print(output2.format(mkCap2))
print(output2.format(mkCap3))
```

```python
# I added Alternative Syntax
sales_amt = 2000
gross_mg_amt = 0.3
print(f'Sales: {sales_amt}, Gross Margin {gross_mg_amt:.2%}')

output2 = ""
mkCaps = [1234, 3456700, 100234]
print(f'Market Cap: {mkCaps[0]:>9,}')
print(f'Market Cap: {mkCaps[1]:>9,}')
print(f'Market Cap: {mkCaps[2]:>9,}')
```

#### Lists and List Methods
Slides 26-32

```python
ticker1 = 'AAPL'
ticker2 = 'SP500'
ticker3 = 'MSFT'

tickers = ['AAPL', 'SP500', 'MSFT', 'TSLA', 'NFLX']
tickers
```

```python
#extract info from a list by ordinal index
tickers[3]
```

```python
#change one of the items
#NFLX --> AMZN
tickers[4] = 'AMZN' #don't forget counting starts at 0
len(tickers) #size of the array --> 5
```

```python
tickers[-1] = 'IBM' #<-- this changes the last item
tickers.append('NFLX') # insert new member at the end
tickers
```

```python
# pull out the first 3 items
tickers[0:3] #0 <= x < 3, 2nd number is open interval
tickers[4:6] #rows 4 and 5
```

#### List Methods
Slide 29

```python
listX = [1, 2, 3]
listY = [100, 200, 300]
listZ = ['a', 'b', 'c']

z = 10

listX.insert(1, z) #adds z as the new second item
listX.append(listY) #adds listY as one entry
listX[4][1] #extracts the "200"

listX.extend(listZ) #breaks appart listZ as multiple "rows"
```

#### Copying Lists

```python
list1 = [1,2,3]
list2 = list1 #Creates a link so both list1 and list2 refer to the same object in memory
list2[0] = 100 #this will change first item of both list1 and list2

list3 = list1.copy() #copies the Values
list4 = list1[:] #copies the Values

list4[0] = 5000 #only changes first item in list4
```
